```python
import math
import numpy as np
#definisco sqrt come radice quadrata
sqrt = math.sqrt

#input
m = int(input("Inserisci il primo numero: "))
n = int(input("Inserisci il secondo numero: "))

def MCD(m,n):
    mcd = 0
    for i in range(1, sqrt(n)):
        if n % i == 0:
            if m % (n//i) == 0:
                return n//i
        elif m % i == 0:
            mcd = i
    return mcd


mcd = MCD(m,n)

print("Il massimo comune divisore tra", m, "e", n, "è:", mcd)
```
